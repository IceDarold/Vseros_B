```python
# Бутстрап Kaggle окружения
import os, pathlib, time, subprocess
from typing import Optional
WORKDIR = pathlib.Path('/kaggle/working')
REPO_URL = 'https://github.com/IceDarold/Vseros_B.git'
REPO_DIR = WORKDIR / 'Vseros_B'

if not REPO_DIR.exists():
    !git clone $REPO_URL $REPO_DIR
os.chdir(REPO_DIR)
os.environ['PYTHONPATH'] = str(REPO_DIR / 'src')
os.environ.setdefault('VPERIOD_RUN_ID', time.strftime('%Y%m%d-%H%M%S'))
RUN_ID = os.environ['VPERIOD_RUN_ID'] + '_exp105'

try:
    from kaggle_secrets import UserSecretsClient  # type: ignore
    wandb_key: Optional[str] = UserSecretsClient().get_secret('WANDB_API_KEY')
except Exception:
    wandb_key = None
if wandb_key:
    os.environ['WANDB_API_KEY'] = wandb_key
    subprocess.run(['wandb', 'login', wandb_key, '--relogin'], check=False)
else:
    print('WARN: WANDB_API_KEY не найден (kaggle_secrets), запускаем в offline-режиме')
    os.environ['WANDB_MODE'] = 'offline'
```
